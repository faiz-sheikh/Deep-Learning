```python
#MNIST MODEL IMPORT AND USAGE
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Loading MNIST dataset
(x_train,y_train), (x_test,y_test) = keras.datasets.mnist.load_data()
```

```python
x_train [0]
plt.imshow(x_train[0])
```

```python
x_train [0]
```

```python
x_train=x_train/255
x_test=x_test/255
```

```python
x_train[0]
```

```python
model = Sequential()
model.add(Flatten(input_shape=(28,28))) #It will convert 28 arrays into single array
model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary ()
```

```python
model.compile(loss='sparse_categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
```

Early Stopping(To Search)

```python
#Model Fitting
history=model.fit(x_train,y_train, epochs=30, validation_split=0.2, batch_size=320)
```

```python
#Testing Model
model.predict(x_test)
y_prob=model.predict(x_test)
print(y_prob)
```

```python
y_pred=y_prob.argmax(axis=1)
print(y_pred)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
accuracy_score(y_test,y_pred)
confusion_matrix(y_test,y_pred)
```

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('Model Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend(['train','validation'])
plt.show()
```

### Trying Early Stopping from Callback Function

```python
#Trying Early Stopping from Callback Function
from tensorflow.keras.callbacks import EarlyStopping
early_stopping=EarlyStopping(monitor='val_loss',patience=3)
history=model.fit(x_train,y_train, epochs=30, validation_split=0.2, batch_size=320,callbacks=[early_stopping])
y_prob=model.predict(x_test)
y_pred=y_prob.argmax(axis=1)
accuracy_score(y_test,y_pred)
confusion_matrix(y_test,y_pred)
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('Model Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend(['train','validation'])
plt.show()
```
